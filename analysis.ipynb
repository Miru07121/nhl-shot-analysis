## Interpretation
Pearson r = −0.0299; Spearman r = −0.0202. There is little linear or monotonic association across teams.

Removing one team at a time gives r from −0.0706 to +0.0182. Carolina leads in volume; Dallas in average quality.

This is descriptive, not causal. Score state and opponent effects are not controlled for. The metrics share an attempt count. xG/60 = volume × quality is an arithmetic identity. See README.md for provenance and limitations.
